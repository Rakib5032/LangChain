```python
from langchain_openai import OpenAIEmbeddings
from langchain_community.vectorstores import Chroma
```

```python
from langchain_core.documents import Document

# Create LangChain documents for Bangladesh cricket players

doc1 = Document(
    page_content="Shakib Al Hasan is regarded as one of the greatest all-rounders in cricket history. Known for his left-handed batting and accurate spin bowling, he has been a key player for Bangladesh in all formats.",
    metadata={"team": "Bangladesh National Cricket Team"}
)

doc2 = Document(
    page_content="Tamim Iqbal is one of the most successful opening batsmen for Bangladesh. His aggressive batting style and consistency at the top order made him a dependable player for many years.",
    metadata={"team": "Bangladesh National Cricket Team"}
)

doc3 = Document(
    page_content="Mashrafe Mortaza is one of the most respected captains in Bangladesh cricket history. Known for his leadership, fighting spirit, and fast bowling, he inspired the team to many memorable victories.",
    metadata={"team": "Bangladesh National Cricket Team"}
)

doc4 = Document(
    page_content="Mustafizur Rahman is a talented left-arm fast bowler famous for his cutters and variations. His performances in T20 and ODI cricket earned him international recognition.",
    metadata={"team": "Bangladesh National Cricket Team"}
)

doc5 = Document(
    page_content="Mushfiqur Rahim is an experienced wicketkeeper-batsman known for his determination and match-winning innings. He has played many crucial roles for Bangladesh in international cricket.",
    metadata={"team": "Bangladesh National Cricket Team"}
)
```

```python
docs = [doc1, doc2, doc3, doc4, doc5]
```

```python
print(docs)
```

```python
from dotenv import load_dotenv
import os

load_dotenv()

api_key = os.getenv("OPENAI_API_KEY")
```

```python
from langchain_openai import OpenAIEmbeddings
embeddings = OpenAIEmbeddings(
    model="text-embedding-3-large",
    base_url="https://openrouter.ai/api/v1",
    api_key=api_key
)
```

```python
from langchain_community.vectorstores import Chroma

vector_store = Chroma(
    embedding_function=embeddings,
    persist_directory="my_chroma_db",
    collection_name="sample"
)
```

```python
vector_store.add_documents(docs)
```

```python
results = vector_store.similarity_search(
    query="who is the good leader among of them?",
    k=2
)
print(results)
```

```python
vector_store.similarity_search_with_score(
    query='Who is sakib?',
    k = 1
)
```

```python
vector_store.similarity_search_with_score(
    query='',
    filter = {'team': 'Bangladesh National Cricket Team'}
)
```

```python
updated_doc4 = Document(
    page_content="Mustafizur Rahman is one of Bangladesh's finest fast bowlers, widely recognized for his deadly cutters, slower deliveries, and excellent performance in pressure situations. He has represented Bangladesh in all formats and played in various international T20 leagues.",
    metadata={"team": "Bangladesh National Cricket Team"}
)

vector_store.update_document(document_id='3a19873c-7ec6-4a00-a27a-dc6b07ac12f0', document=updated_doc4)
```
